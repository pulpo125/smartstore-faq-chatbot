## RAG - retrieve 예제
- 기본 검색 함수 예시 입니다.
- 검색 후 컨텍스트를 만듭니다.

```python
from src.utils import get_chroma_db_client
from src.db.chroma_db import get_or_create_collection
from src.config import cfg_engine
from src.rag.retrieve import retrieve, get_contexts

client = get_chroma_db_client()
collection = get_or_create_collection(
    db_client=client, collection_name=cfg_engine.chroma_db.collection_name
)

doc_metadatas = retrieve(query="가입 서류 어떻게 재출해?", collection=collection)
contexts = get_contexts(doc_metadatas)
print(contexts)
```
